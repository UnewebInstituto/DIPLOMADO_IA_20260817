```python
from sentence_transformers import SentenceTransformer
```

```python
import numpy as np
```

```python
modelo = SentenceTransformer("paraphrase-multilingual-MiniLM-L12-v2")
```

```python
frases = [
    "¿Cómo restablezco mi contraseña?",
    "Olvidé mi clave de acceso",
    "El clima de hoy es soleado",
]
```

```python
frases
```

```python
vectores = modelo.encode(frases, normalize_embeddings=True)
```

```python
vectores
```

```python
print(vectores.shape)  # (3, 384)
```

```python
# Con vectores normalizados, el producto punto es la similitud coseno
print("1 vs 2:", float(np.dot(vectores[0], vectores[1])))  # alta
print("1 vs 3:", float(np.dot(vectores[0], vectores[2])))  # baja
```

```python
print("2 vs 3:", float(np.dot(vectores[1], vectores[2])))  # baja
```

```python
print("1 vs 1:", float(np.dot(vectores[0], vectores[0])))  # alta
```
